#### 7) Convert Likert Scale

```python
# Covert likert Scale into numbers


likert = {
    "Strongly Disagree":1,
    "Disagree":2,
    "Neutral":3,
    "Agree":4,
    "Strongly Agree":5
}

cols = [
    'Transactions_Easier',
    'Purchase_Frequency_Increase',
    'Spend_More_than_Cash',
    'Cashback_Encourages_Spending',
    'Helps_Track_Spending',
    'Supports_Cashless_Economy'
]

for col in cols:
    df[col] = df[col].map(likert)
```

```python
# Mean Values

df[cols].mean()
```

```python
# Correlation Analysis

import seaborn as sns
import matplotlib.pyplot as plt

corr = df[cols].corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
# Price Segmentation

def segment(x):
    if "Small" in str(x):
        return "Low"
    elif "Medium" in str(x):
        return "Medium"
    else:
        return "High"

df['Price_Segment'] = df['Purchase_Type'].apply(segment)

df['Price_Segment'].value_counts()
```

```python
# Compare Spending

df.groupby('Price_Segment')['Spend_More_than_Cash'].mean()
```

#### 8) Hypothesis Testing

```python
# Correlation + p-value

from scipy.stats import pearsonr

# H1
corr1, p1 = pearsonr(df['Transactions_Easier'], df['Spend_More_than_Cash'])

# H2
corr2, p2 = pearsonr(df['Purchase_Frequency_Increase'], df['Spend_More_than_Cash'])

# H3
corr3, p3 = pearsonr(df['Cashback_Encourages_Spending'], df['Spend_More_than_Cash'])

print("H1:", corr1, p1)
print("H2:", corr2, p2)
print("H3:", corr3, p3)
```

```python
# T-Test

# Gender vs Spending

from scipy.stats import ttest_ind

male = df[df['Gender']=="Male"]['Spend_More_than_Cash']
female = df[df['Gender']=="Female"]['Spend_More_than_Cash']

t_stat, p_val = ttest_ind(male, female)

print("T-test p-value:", p_val)
```
